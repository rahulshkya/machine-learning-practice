```python

class PersonalAssistant:

    def __init__(self):
        print("Hi, I am your AI assistant. How can I help?")
   
    def ans_query(self, question=None):

        if question is None:
            question = input("Ask me anything: ")

        response = client.chat.completions.create(
            model="meta-llama/llama-4-scout-17b-16e-instruct",
            messages=[
                {"role": "system", "content": "Act like a helpful personal assistant."},
                {"role": "user", "content": question}
            ],
            temperature=0.7,
            max_completion_tokens=512
        )

        answer = response.choices[0].message.content
        print(answer)
        return answer.strip()

        #summarize email
    def summarize_email(self):

        
        email = input("paste your email: ")
        prompt=f"summarize the following email in 2-3 sentences: {email}"
 
        response = client.chat.completions.create(
            model="meta-llama/llama-4-scout-17b-16e-instruct",
            messages=[
                {"role": "system", "content": "act like an expert email assistent."},
                {"role": "user", "content": prompt}
            ],
            temperature=0.3,
            max_completion_tokens=512
        )
       
        answer = response.choices[0].message.content
        print( "Summary : ",answer)
        return answer.strip()
```

```python
import base64

#images => base64 encoded strings => decode
response = client.chat.completions.create(
            model="meta-llama/llama-4-scout-17b-16e-instruct",
            input="Generate an image of a highly detailed futuristic spaceship , flying near a nebula",
            type=[{"type":"image_generation"}]
        )
image_data=[
    for output in response.output
     if output.type == "image_generation_call"
]

#decode & store image in a file
if image_data:
    image_base64=image_data[0]
    with open("spaceship.png","wb") as f:
        image_bytes=base64.b64decode(image_base64)
        f.write(image_bytes)
```
